```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
plt.rcParams['xtick.direction'] = 'in'
plt.rcParams['ytick.direction'] = 'in'
plt.rcParams['axes.grid'] = True
plt.rcParams['axes.axisbelow'] = True
plt.rcParams['figure.dpi'] = 150
```

```python
p = lambda x : np.exp(-(x+2) * (x+2)) / np.sqrt(np.pi)
q = lambda x : np.exp(-(x-2) * (x-2)) / np.sqrt( np.pi)

fig, ax = plt.subplots()

x = np.linspace(-5, 5, 500)
ax.plot(x, p(x), color='black')
ax.plot(x, q(x), color='red')

ax.set(xlabel='$x$', xticks=[], yticks=[])
ax.text(-2.5, 0.5, '$p(x)$', va='center', ha='right')
ax.text(2.5, 0.5, '$q(x)$', va='center', ha='left', color='red')
ax.spines[['right', 'top']].set_visible(False);
```
